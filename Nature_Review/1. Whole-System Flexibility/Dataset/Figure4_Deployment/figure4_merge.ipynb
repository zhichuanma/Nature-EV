```python
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from pathlib import Path
import fitz  # PyMuPDF (works without Poppler)
```

```python
# Input PDFs
files = {
    "global": "global_deployment.pdf",
    "storage": "storage_deployment.pdf",
    "gas": "gas_deployment.pdf"
}
```

```python



# Convert first page of each PDF to image
images = {}
for name, path in files.items():
    doc = fitz.open(path)
    pix = doc[0].get_pixmap(dpi=150)
    img = pix.tobytes("png")
    images[name] = plt.imread(Path(path).with_suffix(".png"), format="png")
    with open(Path(path).with_suffix(".png"), "wb") as f:
        f.write(img)

# Build 2x2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# [1,1] empty
axes[0, 0].axis("off")

# [1,2] global
axes[0, 1].imshow(images["global"])
axes[0, 1].axis("off")
axes[0, 1].set_title("Global Deployment", fontsize=12)

# [2,1] storage
axes[1, 0].imshow(images["storage"])
axes[1, 0].axis("off")
axes[1, 0].set_title("Storage Deployment", fontsize=12)

# [2,2] gas
axes[1, 1].imshow(images["gas"])
axes[1, 1].axis("off")
axes[1, 1].set_title("Gas Deployment", fontsize=12)

plt.tight_layout()

# Save combined figure as PDF
plt.savefig("merged_deployment.pdf", format="pdf", bbox_inches="tight")
plt.show()

```
